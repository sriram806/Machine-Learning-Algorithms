```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
```

```python
data = {
    "Experiance": [1,2,3,4,5],
    "Salary": [25000,30000,36000,42000,50000]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python

x = df[["Experiance"]]
y = df[["Salary"]]


print(x)
print(y)
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    x,y,test_size=0.2,random_state=42
)

print("Train X Data: ",X_train),
print("Test X Data: ",X_test)
print("Train Y Data: ",y_train)
print("Test Y Data: ",y_test)
```

```python

model = LinearRegression()
model.fit(X_train,y_train)

p = model.predict(X_test)

print("Slope (Coefficient): ", model.coef_[0])
print("Intercept: ", model.intercept_)
print("R2 Score: ", r2_score(y_test,p))
print("MSE: ", mean_squared_error(y_test, p))
print(model.rank_)
```
