```python
plt.figure(figsize=(8,5))

plt.hist(
    df['Temperature'],
    bins=10,
    edgecolor='black'
)

plt.title("Distrubution of Temperatur")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["IceCreamSales"],
    bins=10,
    edgecolor="black"
)

plt.title("Distrubution of IceCream Sales")
plt.xlabel("IceCream Sales")
plt.ylabel("Frequency")

plt.grid(alpha=.3)

plt.tight_layout()
plt.show
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Temperature'],
    df['IceCreamSales'],
    color='blue',
    s=60
)

plt.title("Temperature Vs IceCream Sales")
plt.xlabel("Temperature")
plt.ylabel("Icecream Sales")

plt.grid(alpha=.3)
plt.tight_layout()
plt.show()
```
